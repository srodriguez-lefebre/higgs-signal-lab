# 04 - CV P01 + Random Forest baseline

Baseline robusto. No se tunea ni se prioriza; sirve para dejar una referencia estable contra modelos fuertes.

```python
from __future__ import annotations

import sys
from pathlib import Path

from IPython.display import display
from sklearn.ensemble import RandomForestClassifier

stage_dir = Path.cwd()
if stage_dir.name != "52. Validacion":
    stage_dir = Path.cwd() / "52. Validacion"
sys.path.insert(0, str(stage_dir))

from cv_validation_utils import SEED, run_cv_validation
```

```python
LOG_TO_COMET = True
model = RandomForestClassifier(
    n_estimators=400,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    bootstrap=True,
    class_weight="balanced_subsample",
    random_state=SEED,
    n_jobs=-1,
)
```

```python
fold_metrics, oof_scores, summary = run_cv_validation(
    candidate_name="candidato_4_p01_random_forest_baseline",
    pipeline_id="P01",
    model_name="RandomForestClassifier",
    model=model,
    model_params=model.get_params(deep=True),
    seed=SEED,
    log_to_comet=LOG_TO_COMET,
)
```

```python
display(fold_metrics.round(6))
display(summary.round(6))
```
