# 03 - CV P04 + MLP simple

Candidato a ensemble futuro. No se tunea: se valida estabilidad, OOF scores y thresholds por fold.

```python
from __future__ import annotations

import sys
from pathlib import Path

from IPython.display import display
from sklearn.neural_network import MLPClassifier

stage_dir = Path.cwd()
if stage_dir.name != "52. Validacion":
    stage_dir = Path.cwd() / "52. Validacion"
sys.path.insert(0, str(stage_dir))

from cv_validation_utils import SEED, run_cv_validation
```

```python
LOG_TO_COMET = True
model = MLPClassifier(
    hidden_layer_sizes=(64, 32),
    activation="relu",
    solver="adam",
    alpha=1e-4,
    batch_size=512,
    learning_rate="adaptive",
    learning_rate_init=1e-3,
    max_iter=80,
    shuffle=True,
    random_state=SEED,
    tol=1e-4,
    early_stopping=True,
    validation_fraction=0.1,
    n_iter_no_change=10,
    verbose=False,
)
```

```python
fold_metrics, oof_scores, summary = run_cv_validation(
    candidate_name="candidato_3_p04_mlp_simple",
    pipeline_id="P04",
    model_name="MLPClassifier_simple",
    model=model,
    model_params=model.get_params(deep=True),
    seed=SEED,
    log_to_comet=LOG_TO_COMET,
)
```

```python
display(fold_metrics.round(6))
display(summary.round(6))
```
