# Bloque A - Baselines simples

Experimentos:
- `P00` + Logistic Regression
- `P01` + Logistic Regression
- `P02` + Logistic Regression

Reglas: usa solo datasets ya procesados, no modifica preprocesamiento, no usa `Weight` como feature, registra pipeline/modelo/hiperparametros/seed, reporta AUC, AMS, threshold AMS y tiempo de entrenamiento. Cada corrida se registra en Comet con `project_name="proyecto-1-taa"` y `workspace="equipo-a"`.

```python
from __future__ import annotations

import sys
from pathlib import Path

from IPython.display import display
from sklearn.linear_model import LogisticRegression

stage_dir = Path.cwd()
if stage_dir.name != "51. base":
    stage_dir = Path.cwd() / "51. base"
sys.path.insert(0, str(stage_dir))

from base_experiment_utils import SEED, save_comparison, run_experiment
```

```python
BLOCK_NAME = "bloque_a_baselines_simples"
LOG_TO_COMET = True

def make_logistic_regression() -> LogisticRegression:
    return LogisticRegression(
        penalty="l2",
        C=1.0,
        fit_intercept=True,
        class_weight="balanced",
        solver="saga",
        max_iter=1000,
        tol=1e-4,
        random_state=SEED,
        n_jobs=-1,
    )
```

```python
rows = []

for pipeline_id in ["P00", "P01", "P02"]:
    model = make_logistic_regression()
    row = run_experiment(
        block_name=BLOCK_NAME,
        pipeline_id=pipeline_id,
        model_name="LogisticRegression",
        model=model,
        model_params=model.get_params(deep=True),
        seed=SEED,
        log_to_comet=LOG_TO_COMET,
    )
    rows.append(row)
    print(f"{pipeline_id}: AMS={row['best_ams']:.6f}, AUC={row['roc_auc']:.6f}, threshold={row['best_threshold']:.6f}, fit_seconds={row['fit_seconds']:.2f}")
```

```python
comparison = save_comparison(rows, "01_bloque_a_baselines_simples.csv")
display(
    comparison[
        [
            "pipeline",
            "model",
            "best_ams",
            "roc_auc",
            "best_threshold",
            "precision_at_ams",
            "recall_at_ams",
            "f1_at_ams",
            "fit_seconds",
        ]
    ].round(6)
)
```
