# Bloque D - Redes neuronales

Experimentos:
- `P02` + MLP simple
- `P04` + MLP simple

Reglas: usa solo datasets ya procesados, no modifica preprocesamiento, no usa `Weight` como feature, registra pipeline/modelo/hiperparametros/seed, reporta AUC, AMS, threshold AMS y tiempo de entrenamiento. Cada corrida se registra en Comet con `project_name="proyecto-1-taa"` y `workspace="equipo-a"`.

```python
from __future__ import annotations

import sys
from pathlib import Path

from IPython.display import display
from sklearn.neural_network import MLPClassifier

stage_dir = Path.cwd()
if stage_dir.name != "51. base":
    stage_dir = Path.cwd() / "51. base"
sys.path.insert(0, str(stage_dir))

from base_experiment_utils import SEED, save_comparison, run_experiment
```

```python
BLOCK_NAME = "bloque_d_redes_neuronales"
LOG_TO_COMET = True

def make_mlp_simple() -> MLPClassifier:
    return MLPClassifier(
        hidden_layer_sizes=(64, 32),
        activation="relu",
        solver="adam",
        alpha=1e-4,
        batch_size=512,
        learning_rate="adaptive",
        learning_rate_init=1e-3,
        max_iter=80,
        shuffle=True,
        random_state=SEED,
        tol=1e-4,
        early_stopping=True,
        validation_fraction=0.1,
        n_iter_no_change=10,
        verbose=False,
    )
```

```python
rows = []

for pipeline_id in ["P02", "P04"]:
    model = make_mlp_simple()
    row = run_experiment(
        block_name=BLOCK_NAME,
        pipeline_id=pipeline_id,
        model_name="MLPClassifier_simple",
        model=model,
        model_params=model.get_params(deep=True),
        seed=SEED,
        log_to_comet=LOG_TO_COMET,
    )
    rows.append(row)
    print(f"{pipeline_id}: AMS={row['best_ams']:.6f}, AUC={row['roc_auc']:.6f}, threshold={row['best_threshold']:.6f}, fit_seconds={row['fit_seconds']:.2f}")
```

```python
comparison = save_comparison(rows, "04_bloque_d_redes_neuronales.csv")
display(
    comparison[
        [
            "pipeline",
            "model",
            "best_ams",
            "roc_auc",
            "best_threshold",
            "precision_at_ams",
            "recall_at_ams",
            "f1_at_ams",
            "fit_seconds",
        ]
    ].round(6)
)
```
