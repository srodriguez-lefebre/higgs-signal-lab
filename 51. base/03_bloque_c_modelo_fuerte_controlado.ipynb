# Bloque C - Modelo fuerte controlado

Experimentos:
- `P01` + XGBoost base
- `P02` + XGBoost base
- `P04` + XGBoost base

Reglas: usa solo datasets ya procesados, no modifica preprocesamiento, no usa `Weight` como feature, registra pipeline/modelo/hiperparametros/seed, reporta AUC, AMS, threshold AMS y tiempo de entrenamiento. Cada corrida se registra en Comet con `project_name="proyecto-1-taa"` y `workspace="equipo-a"`.

```python
from __future__ import annotations

import sys
from pathlib import Path

import numpy as np
from IPython.display import display
from xgboost import XGBClassifier

stage_dir = Path.cwd()
if stage_dir.name != "51. base":
    stage_dir = Path.cwd() / "51. base"
sys.path.insert(0, str(stage_dir))

from base_experiment_utils import SEED, save_comparison, run_experiment
```

```python
BLOCK_NAME = "bloque_c_modelo_fuerte_controlado"
LOG_TO_COMET = True

XGB_BASE_PARAMS = {
    "objective": "binary:logistic",
    "eval_metric": "logloss",
    "random_state": SEED,
    "tree_method": "hist",
    "n_jobs": -1,
    "missing": np.nan,
    "learning_rate": 0.05,
    "n_estimators": 400,
    "max_depth": 6,
    "scale_pos_weight": 1.5346,
    "subsample": 0.9,
    "colsample_bytree": 0.9,
    "min_child_weight": 1,
    "gamma": 0.0,
    "reg_lambda": 1.0,
    "reg_alpha": 0.0,
    "max_delta_step": 0,
}

def make_xgboost_base() -> XGBClassifier:
    return XGBClassifier(**XGB_BASE_PARAMS)
```

```python
rows = []

for pipeline_id in ["P01", "P02", "P04"]:
    model = make_xgboost_base()
    row = run_experiment(
        block_name=BLOCK_NAME,
        pipeline_id=pipeline_id,
        model_name="XGBClassifier_base_i3_tree",
        model=model,
        model_params=model.get_params(deep=True),
        seed=SEED,
        log_to_comet=LOG_TO_COMET,
    )
    rows.append(row)
    print(f"{pipeline_id}: AMS={row['best_ams']:.6f}, AUC={row['roc_auc']:.6f}, threshold={row['best_threshold']:.6f}, fit_seconds={row['fit_seconds']:.2f}")
```

```python
comparison = save_comparison(rows, "03_bloque_c_modelo_fuerte_controlado.csv")
display(
    comparison[
        [
            "pipeline",
            "model",
            "best_ams",
            "roc_auc",
            "best_threshold",
            "precision_at_ams",
            "recall_at_ams",
            "f1_at_ams",
            "fit_seconds",
        ]
    ].round(6)
)
```
