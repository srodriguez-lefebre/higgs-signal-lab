# Bloque B - Baseline robusto

Experimentos:
- `P01` + Random Forest
- `P02` + Random Forest

Reglas: usa solo datasets ya procesados, no modifica preprocesamiento, no usa `Weight` como feature, registra pipeline/modelo/hiperparametros/seed, reporta AUC, AMS, threshold AMS y tiempo de entrenamiento. Cada corrida se registra en Comet con `project_name="proyecto-1-taa"` y `workspace="equipo-a"`.

```python
from __future__ import annotations

import sys
from pathlib import Path

from IPython.display import display
from sklearn.ensemble import RandomForestClassifier

stage_dir = Path.cwd()
if stage_dir.name != "51. base":
    stage_dir = Path.cwd() / "51. base"
sys.path.insert(0, str(stage_dir))

from base_experiment_utils import SEED, save_comparison, run_experiment
```

```python
BLOCK_NAME = "bloque_b_baseline_robusto"
LOG_TO_COMET = True

def make_random_forest() -> RandomForestClassifier:
    return RandomForestClassifier(
        n_estimators=400,
        criterion="gini",
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features="sqrt",
        bootstrap=True,
        class_weight="balanced_subsample",
        random_state=SEED,
        n_jobs=-1,
    )
```

```python
rows = []

for pipeline_id in ["P01", "P02"]:
    model = make_random_forest()
    row = run_experiment(
        block_name=BLOCK_NAME,
        pipeline_id=pipeline_id,
        model_name="RandomForestClassifier",
        model=model,
        model_params=model.get_params(deep=True),
        seed=SEED,
        log_to_comet=LOG_TO_COMET,
    )
    rows.append(row)
    print(f"{pipeline_id}: AMS={row['best_ams']:.6f}, AUC={row['roc_auc']:.6f}, threshold={row['best_threshold']:.6f}, fit_seconds={row['fit_seconds']:.2f}")
```

```python
comparison = save_comparison(rows, "02_bloque_b_baseline_robusto.csv")
display(
    comparison[
        [
            "pipeline",
            "model",
            "best_ams",
            "roc_auc",
            "best_threshold",
            "precision_at_ams",
            "recall_at_ams",
            "f1_at_ams",
            "fit_seconds",
        ]
    ].round(6)
)
```
