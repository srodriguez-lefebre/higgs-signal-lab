# 02 - CV P02 + XGBoost base

Representante simple para arboles. Objetivo: comparar si el feature engineering de `P04` mejora de forma estable frente a `P02`, sin tuning.

```python
from __future__ import annotations

import sys
from pathlib import Path

import numpy as np
from IPython.display import display
from xgboost import XGBClassifier

stage_dir = Path.cwd()
if stage_dir.name != "52. Validacion":
    stage_dir = Path.cwd() / "52. Validacion"
sys.path.insert(0, str(stage_dir))

from cv_validation_utils import SEED, run_cv_validation
```

```python
LOG_TO_COMET = True
MODEL_PARAMS = {
    "objective": "binary:logistic",
    "eval_metric": "logloss",
    "random_state": SEED,
    "tree_method": "hist",
    "n_jobs": -1,
    "missing": np.nan,
    "learning_rate": 0.05,
    "n_estimators": 400,
    "max_depth": 6,
    "scale_pos_weight": 1.5346,
    "subsample": 0.9,
    "colsample_bytree": 0.9,
    "min_child_weight": 1,
    "gamma": 0.0,
    "reg_lambda": 1.0,
    "reg_alpha": 0.0,
    "max_delta_step": 0,
}
model = XGBClassifier(**MODEL_PARAMS)
```

```python
fold_metrics, oof_scores, summary = run_cv_validation(
    candidate_name="candidato_2_p02_xgboost_base",
    pipeline_id="P02",
    model_name="XGBClassifier_base",
    model=model,
    model_params=model.get_params(deep=True),
    seed=SEED,
    log_to_comet=LOG_TO_COMET,
)
```

```python
display(fold_metrics.round(6))
display(summary.round(6))
```
