# 03 - Comparacion tuning XGBoost

Lee los summaries generados por los notebooks `01` y `02` y arma una comparacion final entre `P04` y `P02` para `T01`-`T04`.

```python
from __future__ import annotations

from pathlib import Path

import pandas as pd
from IPython.display import display


def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "roadmap.md").exists() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("No se pudo localizar la raiz del proyecto.")


project_root = find_project_root()
results_dir = project_root / "53. Tuning de XGBoost" / "results"
```

```python
summary_paths = [
    results_dir / "p04_xgboost_lr_estimators_summary.csv",
    results_dir / "p02_xgboost_lr_estimators_summary.csv",
]
missing = [path for path in summary_paths if not path.exists()]
if missing:
    raise FileNotFoundError(f"Faltan summaries. Ejecutar primero notebooks 01 y 02: {missing}")

comparison = pd.concat([pd.read_csv(path) for path in summary_paths], ignore_index=True)
comparison = comparison.sort_values(["ams_mean", "ams_std", "fit_seconds_mean"], ascending=[False, True, True]).reset_index(drop=True)
comparison_path = results_dir / "03_comparacion_tuning_xgboost.csv"
comparison.to_csv(comparison_path, index=False)

display(
    comparison[
        [
            "pipeline",
            "config_id",
            "learning_rate",
            "n_estimators",
            "ams_mean",
            "ams_std",
            "auc_mean",
            "auc_std",
            "threshold_median",
            "threshold_std",
            "signal_rate_mean",
            "precision_at_ams_mean",
            "recall_at_ams_mean",
            "f1_at_ams_mean",
            "fit_seconds_mean",
        ]
    ].round(6)
)
print(f"Comparacion guardada en: {comparison_path}")
```

## Decision

Elegir por balance, no por maximo aislado: AMS medio alto, AMS std bajo, tiempo razonable y threshold estable.
