# 02 - Comparacion sampling tuning

Ordena los resultados de `subsample` y `colsample_bytree` por balance: AMS medio, estabilidad, tiempo y threshold/signal rate razonables.

```python
from __future__ import annotations

from pathlib import Path

import pandas as pd
from IPython.display import display


def find_project_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "roadmap.md").exists() and (candidate / "data").is_dir():
            return candidate
    raise FileNotFoundError("No se pudo localizar la raiz del proyecto.")


project_root = find_project_root()
results_dir = project_root / "55. Tuning de subsample y colsample_bytree" / "results"
```

```python
summary_path = results_dir / "p04_t03_sampling_summary.csv"
if not summary_path.exists():
    raise FileNotFoundError(f"Falta summary. Ejecutar primero notebook 01: {summary_path}")

comparison = pd.read_csv(summary_path)
comparison = comparison.sort_values(["ams_mean", "ams_std", "fit_seconds_mean"], ascending=[False, True, True]).reset_index(drop=True)
comparison_path = results_dir / "02_comparacion_sampling.csv"
comparison.to_csv(comparison_path, index=False)

display(
    comparison[
        [
            "sampling_config_id",
            "subsample",
            "colsample_bytree",
            "ams_mean",
            "ams_std",
            "auc_mean",
            "auc_std",
            "threshold_median",
            "signal_rate_mean",
            "precision_at_ams_mean",
            "recall_at_ams_mean",
            "f1_at_ams_mean",
            "fit_seconds_mean",
        ]
    ].round(6)
)
print(f"Comparacion guardada en: {comparison_path}")
```
